# Extract all ZIP datasets


```python
import zipfile
from pathlib import Path

EXTRACT_DIR = DATASET_DIR / "Extracted"
EXTRACT_DIR.mkdir(exist_ok=True)

zip_files = list(DATASET_DIR.glob("*.zip"))

for zip_path in zip_files:

    dataset_name = zip_path.stem
    extract_to = EXTRACT_DIR / dataset_name

    if extract_to.exists():
        print(f"✅ {dataset_name} already extracted.")
        continue

    print(f"\nExtracting {dataset_name} ...")

    extract_to.mkdir(exist_ok=True)

    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_to)

    print(f"Finished extracting {dataset_name}")

print("\n All datasets extracted!")
```

```python
coco_zip = DATASET_DIR / "coco2017.zip"
extract_dir = DATASET_DIR / "Extracted" / "coco2017"
extract_dir.mkdir(parents=True, exist_ok=True)

print("Extracting only required COCO files...")

with zipfile.ZipFile(coco_zip, "r") as z:

    for member in z.infolist():

        name = member.filename

        # Extract ONLY these folders
        if (
            name.startswith("coco2017/train2017/")
            or name.startswith("coco2017/val2017/")
            or name.startswith("coco2017/annotations/")
        ):
            z.extract(member, extract_dir)

print("COCO extraction completed.")
```
