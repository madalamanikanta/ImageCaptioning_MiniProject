# Step 2 : Import Required Libraries


```python
from pathlib import Path
import pandas as pd
import re
import string
```

# Step 3 : Define Project Paths


```python
PROJECT_DIR = Path("/content/drive/MyDrive/ImageCaptioning_MiniProject")

PROCESSED_DIR = PROJECT_DIR / "Processed"

merged_file = PROCESSED_DIR / "merged_dataset.csv"
```

# Step 4 : Load Merged Dataset


```python
df = pd.read_csv(merged_file)

print(df.shape)

df.head()
```

# Step 5 : Dataset Inspection


```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df["dataset"].value_counts())
```

```python
# Create a copy of the dataset
clean_df = df.copy()

print("Original Dataset Size :", len(clean_df))

# Remove rows with missing captions
clean_df = clean_df.dropna(subset=["caption"]).reset_index(drop=True)

print("Dataset Size After Removing Missing Captions :", len(clean_df))

print("\nMissing Captions Remaining :")

print(clean_df["caption"].isnull().sum())
```

## Step 6 : Create a Backup of the Original Dataset

```python
# Create a copy of the dataset

clean_df = df.copy()
print("Backup created successfully!")
print(clean_df.head())
```

Step 7 : Convert Captions to Lowercase

```python
clean_df["caption"] = clean_df["caption"].str.lower()

print("Lowercase conversion completed!")

clean_df.head()
```

##Step 8 : Remove Punctuation

```python
import string

translator = str.maketrans("", "", string.punctuation)

clean_df["caption"] = clean_df["caption"].apply(
    lambda x: x.translate(translator)
)

print("Punctuation removed!")

clean_df.head()
```

##Step 9 : Remove Extra White Spaces

```python
# Remove extra spaces

clean_df["caption"] = clean_df["caption"].apply(
    lambda x: re.sub(r"\s+", " ", x).strip()
)

print("Extra spaces removed successfully!")

clean_df.head()
```

## Step 10 : Remove Empty Captions

```python
before = len(clean_df)

clean_df = clean_df[
    clean_df["caption"].str.strip() != ""
]

after = len(clean_df)

print("Removed Empty Captions :", before - after)

print("Remaining Captions :", after)
```

## Step 11 : Remove Very Short Captions

```python
before = len(clean_df)

clean_df = clean_df[
    clean_df["caption"].apply(
        lambda x: len(x.split()) >= 3
    )
]

after = len(clean_df)

print("Removed Short Captions :", before - after)

print("Remaining Captions :", after)
```

Step 12 : Remove Duplicate Captions

```python
before = len(clean_df)

clean_df = clean_df.drop_duplicates()

after = len(clean_df)

print("Duplicate Rows Removed :", before - after)

print("Remaining Samples :", after)
```

## Step 13 : Add Start and End Tokens

```python
clean_df["caption"] = clean_df["caption"].apply(
    lambda x: f"<start> {x} <end>"
)

print("Special tokens added successfully!")

clean_df.head()
```

## Step 14 : Caption Length Analysis (NEW)

```python
clean_df["caption_length"] = clean_df["caption"].apply(
    lambda x: len(x.split())
)

print(clean_df["caption_length"].describe())
```

## Step 15 : Remove Temporary Column

```python
clean_df.drop(
    columns=["caption_length"],
    inplace=True
)
```

## Step 16 : Save Cleaned Dataset

```python
output_file = PROCESSED_DIR / "cleaned_dataset.csv"

clean_df.to_csv(
    output_file,
    index=False
)

print("Dataset saved successfully!")

print(output_file)
```
